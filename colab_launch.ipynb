# 🎵 Spotify Virality Prediction — Dashboard Launcher
Run each cell in order. The dashboard will open via a public tunnel URL.

```
# ── Step 1: Clone / pull the repo ────────────────────────────────────────────
import os

REPO_URL  = 'https://github.com/black1000u-blip/spotify-viral-ml.git'
REPO_DIR  = 'spotify-viral-ml'

if os.path.exists(REPO_DIR):
    print('Repo already cloned — pulling latest...')
    os.chdir(REPO_DIR)
    os.system('git pull')
else:
    os.system(f'git clone {REPO_URL}')
    os.chdir(REPO_DIR)

print('Working directory:', os.getcwd())
print('Contents:', os.listdir('.'))
```

```
# ── Step 2: Install all dependencies ─────────────────────────────────────────
!pip install -q -r requirements.txt
!pip install -q pyngrok
print('✅ All packages installed.')
```

```
# ── Step 3: Set up ngrok tunnel ───────────────────────────────────────────────
# Get a FREE auth token from https://dashboard.ngrok.com/get-started/your-authtoken
# Paste it below:
NGROK_AUTH_TOKEN = ''   # <-- paste your token here

from pyngrok import ngrok, conf

if NGROK_AUTH_TOKEN:
    conf.get_default().auth_token = NGROK_AUTH_TOKEN
    public_url = ngrok.connect(8501)
    print(f'\n🌐 Dashboard URL: {public_url}')
    print('Open the URL above in your browser\n')
else:
    print('⚠️  No ngrok token provided!')
    print('Get one free at: https://dashboard.ngrok.com/get-started/your-authtoken')
    print('Then paste it into NGROK_AUTH_TOKEN above and re-run.')
```

```
# ── Step 4: Launch the Streamlit app ─────────────────────────────────────────
# This cell runs the server. Keep it running — the dashboard stays live!
# The URL printed in Step 3 is your public link.
!python -m streamlit run frontend/app.py \
    --server.port 8501 \
    --server.headless true \
    --server.enableCORS false \
    --server.enableXsrfProtection false
```
